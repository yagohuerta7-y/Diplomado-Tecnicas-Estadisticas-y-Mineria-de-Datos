## Ejercicio 1.8: Tipos de datos y operadores

Dada una cadena de texto `frase`, devuelve `True` si es un palíndromo; en caso contrario, devuelve `False`. Un palíndromo es una cadena que se lee igual de izquierda a derecha que de derecha a izquierda. Además, no distingue entre mayúsculas y minúsculas, y ignora todos los caracteres que no sean letras ni números.

El método `.isalnum()` en Python se utiliza para verificar si todos los caracteres de una cadena son alfanuméricos, es decir:

- Letras (mayúsculas o minúsculas): A-Z, a-z
- Números: 0-9

````python
def es_palindromo(frase: str) -> bool:
    """Devuelve True si `frase` es un palíndromo, ignorando mayúsculas y caracteres no alfanuméricos."""
    limpio = "".join(caracter for caracter in frase if caracter.isalnum()).lower()
    return limpio == limpio[::-1]
````

````python
casos_prueba = [
    "Anita lava la tina",
    "Sé verlas al revés",
    "Hola mundo",
    "A man, a plan, a canal: Panama",
]

for caso in casos_prueba:
    print(f"{caso!r:40} -> {es_palindromo(caso)}")
````

## Ejercicio 1.9: Estructuras de datos

Escribe una función en Python para obtener la intersección de dos listas.

Por ejemplo, si:

```python
A = [1, 2, 3, 4, 5]
B = [0, 1, 3, 7]
```

Entonces la salida debe ser:

```python
[1, 3]
```

````python
def interseccion_listas(lista_1, lista_2):
    interseccion = set(lista_1) & set(lista_2)
    return list(interseccion)
````

````python
A = [1, 2, 3, 4, 5]
B = [0, 1, 3, 7]
interseccion_listas(A, B)
````

## Ejercicio 1.10: Estructuras de control

Escribe una función que evalúe si una persona cumple con los requisitos para acceder a un crédito, considerando su edad, nivel de ingresos y monto de deuda. La función debe devolver un mensaje con el veredicto final, además los datos deberán ser solicitados directamente al usuario a través de la consola función `input()`.

Para que una persona sea considerada apta para recibir el crédito, debe cumplir con todos los siguientes requisitos:

- Edad mínima: Tener 21 años o más.
- Ingresos mensuales: Percibir más de $15,000.
- Relación deuda/ingresos: El porcentaje de deuda respecto al ingreso debe ser menor a 40%.

**Nota:** como esta función pide datos por consola con `input()`, la celda de abajo solo la define, sin ejecutarla, ya que un notebook corrido de forma automática no puede recibir entrada interactiva real. Para probarla, abre el notebook y ejecuta `evaluar_cliente()` en tu propia sesión.

````python
def evaluar_cliente():
    try:
        nombre = input("Ingresa tu nombre: ")
        edad = int(input("Ingresa tu edad: "))
        ingresos = float(input("¿Cuál es tu ingreso mensual? "))
        deuda = float(input("¿A cuánto asciende tu deuda? "))

        if edad >= 21 and ingresos > 15000 and deuda / ingresos < 0.4:
            mensaje = f"{nombre}, eres candidato a un crédito"
        else:
            mensaje = f"{nombre}, crédito denegado"
    except ValueError:
        mensaje = "La edad, el ingreso y la deuda deben de ser valores numéricos"
    return mensaje


# evaluar_cliente()  # descomentar y ejecutar de forma interactiva para probar la función
````

## Ejercicio 1.11: Numpy

Crea un array NumPy de temperaturas diarias (30 elementos).

a) Calcula: media, máximo, mínimo.

b) Aplica la estandarización Z-score.

c) Cuenta los días con temperatura superior a la media.

````python
import numpy as np

rng = np.random.default_rng(seed=42)
temperaturas = rng.uniform(low=20, high=35, size=30)
temperaturas
````

````python
# a) Media, mínimo, máximo y desviación estándar
media = temperaturas.mean()
minimo = temperaturas.min()
maximo = temperaturas.max()
desviacion = temperaturas.std()
media, minimo, maximo, desviacion
````

````python
# b) Estandarización Z-score
z_score = (temperaturas - media) / desviacion
z_score
````

````python
# c) Días con temperatura superior a la media
dias_sobre_media = temperaturas[temperaturas > media].shape[0]
dias_sobre_media
````

## Ejercicio 1.12: Pandas

Dispones de la información del inventario y las ventas de una tienda. Con base en estos datos, elabora un resumen que incluya: qué productos se agotaron, cuáles se vendieron más y cuáles no se vendieron.

````python
import pandas as pd

inventario = {
    "producto": [
        "Agua embotellada", "Refresco de cola", "Jugo de mango",
        "Sopa instantánea", "Sandwich de jamón", "Papas fritas",
        "Pan integral", "Pan dulce", "Galletas",
        "Shampoo en sobre", "Desodorante en barra", "Toallitas húmedas",
    ],
    "categoria": [
        "Bebidas", "Bebidas", "Bebidas",
        "Alimentos preparados", "Alimentos preparados", "Botanas",
        "Panadería", "Panadería", "Panadería",
        "Higiene personal", "Higiene personal", "Higiene personal",
    ],
    "inventario_inicial": [4, 20, 18, 15, 12, 20, 30, 25, 22, 40, 35, 50],
}

ventas = {
    "producto": [
        "Agua embotellada", "Refresco de cola", "Jugo de mango",
        "Sopa instantánea", "Papas fritas",
        "Pan integral", "Pan dulce", "Galletas",
        "Shampoo en sobre", "Desodorante en barra", "Toallitas húmedas",
        "Agua embotellada", "Papas fritas", "Toallitas húmedas",
    ],
    "ventas": [2, 1, 3, 2, 2, 1, 1, 1, 4, 1, 3, 2, 1, 1],
}

df_inventario = pd.DataFrame(inventario)
df_ventas = pd.DataFrame(ventas)
df_inventario
````

````python
df_ventas_totales = df_ventas.groupby("producto")["ventas"].sum().to_frame()
df_ventas_totales
````

````python
df = pd.merge(df_inventario, df_ventas_totales, how="left", on="producto")
df["ventas"] = df["ventas"].fillna(0)
df["inventario_final"] = df["inventario_inicial"] - df["ventas"]
df
````

````python
# Productos que se agotaron
productos_agotados = df[df["inventario_final"] == 0]["producto"].to_list()
productos_agotados
````

````python
# Productos que más se vendieron
df.sort_values(by="ventas", ascending=False).head(5).reset_index(drop=True)[["producto", "ventas"]]
````

````python
# Productos que no se vendieron
productos_no_vendidos = df[df["ventas"] == 0]["producto"].to_list()
productos_no_vendidos
````

## Ejercicio 1.14: SQL

Realiza las siguientes consultas para la tabla `t_customers` de la base de datos `customers.db`:

- **a)** Selecciona todos los registros en los que la columna `City` tenga el valor `Berlin` y la columna `PostalCode` tenga el valor `12209`.
- **b)** Selecciona todos los registros en los que la columna `Country` tenga el valor `Germany` y la columna `City` no tenga el valor `Berlin`.
- **c)** Selecciona todos los registros cuya columna `City` tenga el valor `Madrid` o `Barcelona`.

````python
import sqlite3

DB_PATH = Path("../../Data/customers.db")
assert DB_PATH.exists(), f"No se encontró la base de datos en {DB_PATH.resolve()}"

conn = sqlite3.connect(DB_PATH)
````

### a) Ciudad `Berlin` y código postal `12209`

````python
query_a = """
SELECT *
FROM t_customers
WHERE City = 'Berlin' AND PostalCode = '12209';
"""

pd.read_sql_query(query_a, conn)
````

### b) País `Germany` y ciudad distinta de `Berlin`

````python
query_b = """
SELECT *
FROM t_customers
WHERE Country = 'Germany' AND City != 'Berlin';
"""

pd.read_sql_query(query_b, conn)
````

### c) Ciudad `Madrid` o `Barcelona`

````python
query_c = """
SELECT *
FROM t_customers
WHERE City IN ('Madrid', 'Barcelona');
"""

pd.read_sql_query(query_c, conn)
````
